```python
# Convert to grayscale for filtering
gray = cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
plt.imshow(gray)
```

```python
import numpy as np

# vertical edge detection
sobel_y = np.array([[ -1, -2, -1],
                   [ 0, 0, 0],
                   [ 1, 2, 1]])


# horizontal edge detection
sobel_x = np.array([[-1, 0, 1],
                   [-2, 0, 2],
                   [-1, 0, 1]])

# filter the image using filter2D(grayscale image, bit-depth, kernel)
# -1 means:- Keep the same depth/data type as the input image.
filt_img1 = cv2.filter2D(gray,-1,sobel_x)
filt_img2 = cv2.filter2D(gray,-1,sobel_y)
f,ax = plt.subplots(1,2,figsize=(20,10))
ax[0].set_title('Vertical edge detection')
ax[0].imshow(filt_img1,cmap='gray')
ax[1].set_title('Horizontal edge detection')
ax[1].imshow(filt_img2,cmap='gray')
```
